# Deep Hedging (JAX port)
## Testing the SimpleWorld_Spot_ATM world generator

Simple World with one asset and one floating ATM option.
    The asset has stochastic volatility, and a mean-reverting drift.
    The implied volatility of the asset is not the realized volatility.

Port of `notebooks/simpleWorld_Spot_ATM.ipynb` by Hans Buehler, June 30 2022.

```python
import os, sys
sys.path.insert(0, os.path.abspath(".."))
import numpy as np
import matplotlib.pyplot as plt
from deephedging.world import WorldConfig, simulate
from deephedging.plotting import plot_world
```

## Simple Black & Scholes World

```python
world = simulate(WorldConfig(samples=10000, black_scholes=True, drift=0.15, rvol=0.22))
plot_world(world, plot_samples=10)
plt.show()
```

## Stochastic Vol, Drift, Option trading

```python
world = simulate(WorldConfig(samples=10000, black_scholes=False, invar_steps=0))
plot_world(world, plot_samples=10)
plt.show()
```

## Stochastic Vol, Drift, Option trading; in the 'invariant' distribution

```python
world = simulate(WorldConfig(samples=10000, black_scholes=False, invar_steps=10))
plot_world(world, plot_samples=10)
plt.show()
```
