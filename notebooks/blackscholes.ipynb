# Deep Hedging AI (JAX port)
### Convergence to Black & Scholes Delta

Port of the five notebooks in `notebooks/blackscholes/` by Hans Buehler, January 2nd, 2023: `trainer_bs_delta.ipynb`, `trainer_bs_delta_exp2@10.ipynb`, `trainer_bs_delta_nodrift.ipynb`, `trainer_bs_delta_nodrift_exp2@10.ipynb`, `trainer_bs_small.ipynb`. One section per original notebook.

```python
import os, sys
sys.path.insert(0, os.path.abspath(".."))
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, Markdown
from deephedging.world import WorldConfig, simulate
from deephedging.model import AgentConfig, GymConfig
from deephedging.train import TrainConfig, fit
from deephedging.plotting import plot_training, plot_blackscholes

# DH_EPOCHS overrides every epoch count for quick smoke runs; DH_CHECKPOINTS="" turns checkpoints off.
epochs = lambda n: int(os.environ.get("DH_EPOCHS") or n)
CHECKPOINTS = os.environ.get("DH_CHECKPOINTS", "checkpoints") or None
```

## 1. `trainer_bs_delta.ipynb`
#### Base case: BS with (statistical) drift 0.1, cvar@1, 400 epochs, zero cost, 50000 path
#### This file: base case

Hans Buehler, January 2nd, 2023

### Deep Hedging in a Black & Scholes World

```python
world_cfg = WorldConfig(samples=50000, steps=20, black_scholes=True, rvol=0.2, drift=0.1, cost_s=0.)
gym = GymConfig(agent=AgentConfig(depth=5, width=40, activation="softplus"), utility="cvar", lmbda=1.)
train_cfg = TrainConfig(epochs=epochs(400))

world = simulate(world_cfg)
val_world = world.clone(samples=world.nSamples // 10, seed=1)
base = fit(gym, train_cfg, world, val_world, log_every=50, checkpoint_dir=CHECKPOINTS)
plot_training(base.result, base.train, base.val, world, val_world)
plt.show()
print("Keys of the dictionary returned by the rollout: ", list(base.train))
```

### Hedging results in detail

```python
plot_blackscholes(world, base.train, gym, vol=world_cfg.rvol)
plt.show()
```
